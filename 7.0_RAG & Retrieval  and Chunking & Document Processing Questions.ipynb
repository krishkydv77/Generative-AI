
# Part 1: RAG & Retrieval Questions

## 1. Your RAG system retrieves irrelevant documents even though the correct documents exist in the database. How would you debug this issue?

Answer:

First, I would check whether the correct documents are properly stored and indexed in the vector database.

Then, I would check the chunking strategy, embedding model, and query preprocessing. I would also check the similarity scores and the Top-K results.

If necessary, I would use a reranker to improve the ranking of retrieved documents.

Key points: Check indexing, chunking, embeddings, Top-K, and reranking.

## 2. A user searches for "refund policy," but the documents use the term "money-back procedure." Why might the retriever fail, and how would you fix it?

Answer:

The retriever may fail because the query and document use different words for the same meaning.

A keyword-based retriever may not recognize that both terms are related.

I would use semantic search with embeddings and hybrid search that combines keyword search and vector search. I could also use query expansion to add related terms.

Key points: Semantic search, hybrid search, and query expansion.

## 3. Your system performs well for English documents but poorly for queries in another language. What would you investigate?

Answer:

First, I would check whether the embedding model supports both languages.

Some embedding models perform well in English but poorly in other languages.

I would use a multilingual embedding model and test it with queries and documents in both languages. I would also check language-specific preprocessing and evaluate retrieval quality separately for each language.

Key points: Multilingual embeddings, preprocessing, and language-wise evaluation.

## 4. You changed the embedding model from Model A to Model B. Can you reuse your existing vector index? Why or why not?

Answer:

Usually, I cannot directly reuse the existing vector index.

Different embedding models may produce vectors with different dimensions and different meanings. Even if both models produce vectors of the same dimension, their vector spaces may be different.

I would generate new embeddings for all documents and rebuild the index using Model B.

Key points: Re-embedding and re-indexing are generally required when changing the embedding model.

## 5. The cosine similarity score is high, but the generated answer is still incorrect. What could be causing this issue?

Answer:

A high cosine similarity score does not guarantee that a document contains the correct answer.

The retrieved document may be related to the query but may not contain the required facts. The LLM may also misunderstand the context or generate unsupported information.

I would inspect the retrieved chunks, use a reranker, improve the prompt, and ask the model to answer only from the provided context.

Key points: Relevance is not the same as factual correctness.

## 6. Your database contains 10 billion chunks, and search latency is too high. How would you optimize retrieval performance?

Answer:

For 10 billion chunks, I would use a scalable distributed vector database with approximate nearest neighbor (ANN) search.

I would also use metadata filtering, sharding, quantization, and caching to reduce search time.

If necessary, I would use a two-stage retrieval system: first retrieve a candidate set quickly, then rerank only the best candidates.

I would measure latency and retrieval accuracy after every optimization.

Key points: ANN, distributed indexing, sharding, quantization, caching, and reranking.

## 7. A query contains an exact invoice ID, but semantic search retrieves a similar invoice instead. How would you solve this problem?

Answer:

Semantic search is designed to find similar meanings, not necessarily exact identifiers.

I would use exact matching on the invoice ID through a database query or keyword search.

I would give exact ID matches higher priority and use semantic search only for questions that require understanding the invoice content.

Key points: Exact matching, metadata filtering, and hybrid retrieval.

## 8. You have only 500 domain-specific documents. How would you decide between using a general-purpose embedding model and fine-tuning a custom model?

Answer:

I would start with a general-purpose embedding model because it is easier and cheaper to implement.

Then, I would evaluate it using domain-specific queries and relevant document pairs.

If retrieval performance is poor and I have enough high-quality training examples, I would consider fine-tuning an embedding model.

With only 500 documents, I would not fine-tune automatically because the dataset may be too small for reliable improvements.

Key points: Start with a general model, evaluate it, and fine-tune only when necessary.

## 9. After adding thousands of new documents, retrieval quality decreases. What might be the reasons?

Answer:

New documents may contain duplicate information, irrelevant content, poor-quality text, or incorrect metadata.

They may also introduce similar chunks that compete with the correct results.

I would check document quality, chunking, metadata, embedding consistency, and retrieval metrics. I would also remove duplicates and use metadata filters or reranking where needed.

Key points: Data quality, duplicates, metadata, and index consistency.

## 10. You need both semantic understanding and exact term matching. What retrieval architecture would you propose?

Answer:

I would use a hybrid retrieval architecture.

First, BM25 or another keyword-based retriever would find documents containing important exact terms. At the same time, a vector retriever would find semantically related documents.

Then, I would combine their results using Reciprocal Rank Fusion (RRF) and use a reranker to select the most relevant chunks.

Finally, I would pass those chunks to the LLM.

Key points: BM25 + vector search + RRF + reranking.

# Part 2: Chunking & Document Processing Questions

## 1. Your documents contain long legal clauses, and a fixed-size chunking strategy splits a clause across multiple chunks. What would you change?

Answer:

I would use structure-aware chunking instead of relying only on fixed-size chunks.

I would split documents using legal sections, clauses, and subclauses. I would try to keep each complete clause in one chunk.

If a clause is too long, I would split it carefully and maintain some overlap so that important context is not lost.

Key points: Structure-aware chunking, clause boundaries, and overlap.

## 2. A chunk contains a table with 20 rows, but the user asks a question about only one row. How would you improve retrieval?

Answer:

I would process the table separately instead of treating the entire table as one large text chunk.

I could store each row as a separate record with column names and metadata. For example, an invoice table could be stored with invoice ID, customer name, date, and amount.

Then, I would use exact filtering or metadata search to find the required row.

Key points: Row-level chunking, structured data, and metadata filtering.

## 3. Your chunks are 2,000 tokens long, and the model receives a lot of irrelevant context. What would you do?

Answer:

I would reduce the chunk size and split the documents into smaller, meaningful sections.

I would use structure-aware or recursive chunking to preserve important context. I would also use a reranker to select the most relevant chunks before sending them to the LLM.

I would test different chunk sizes to find the best balance between context and relevance.

Key points: Smaller chunks, meaningful boundaries, and reranking.

## 4. Your chunks are only 50 tokens long, and the retriever does not return enough context to answer the question accurately. How would you fix this?

Answer:

I would increase the chunk size so that each chunk contains more complete information.

I could also increase the overlap between chunks or retrieve neighboring chunks when a relevant chunk is found.

I would evaluate different chunk sizes and overlap values to find the best configuration.

Key points: Larger chunks, overlap, and neighboring-chunk retrieval.

## 5. Your knowledge base contains source code. Why might a generic recursive text splitter perform poorly on code documents?

Answer:

A generic recursive text splitter may split code in the middle of a function, class, or logical block.

This can separate important dependencies and make the code difficult to understand.

I would use code-aware chunking based on programming language syntax, functions, classes, and abstract syntax trees (ASTs) where appropriate.

I would also preserve metadata such as file path, class name, and function name.

Key points: Code-aware chunking, AST, functions, classes, and metadata.

## 6. A document contains repeated headers, footers, and page numbers. How can these affect chunking and embeddings?

Answer:

Repeated headers, footers, and page numbers add noise to the document.

They may make unrelated chunks look similar and reduce retrieval quality.

I would clean the document before chunking by removing repeated headers, footers, and unnecessary page numbers while preserving useful information.

Key points: Document cleaning and noise reduction.

## 7. A user asks a question whose answer spans three consecutive chunks. How would you retrieve the complete information?

Answer:

I would use neighboring-chunk retrieval.

First, I would retrieve the most relevant chunk. Then, I would fetch the previous and next chunks using document IDs and chunk positions.

I would combine the necessary chunks in the correct order and remove duplicate content before passing the context to the LLM.

Key points: Chunk metadata, neighboring chunks, and context reconstruction.

## 8. Your RAG application continuously receives updated documents. How would you manage chunk updates and re-indexing?

Answer:

I would assign a unique document ID and version to every document.

When a document changes, I would identify the changed version, generate new chunks and embeddings, and update the vector index.

I would delete or deactivate the old chunks to prevent duplicate or outdated information from being retrieved.

I would also use a scheduled or event-driven indexing pipeline and maintain logs for failures.

Key points: Document versioning, incremental indexing, and stale-chunk removal.

## 9. Your chunker preserves document headings, but retrieved chunks do not reveal which document they came from. What metadata would you add?

Answer:

I would add metadata such as:

- Document ID and document title
- Source file path or URL
- Section heading and subsection heading
- Page number
- Chunk ID and chunk position
- Document version and update timestamp
- Access permissions, if required

This metadata would help identify the source, retrieve neighboring chunks, filter results, and provide citations.

Key points: Source traceability, filtering, and citations.

## 10. If two vectors have a cosine similarity of 0.95, does that guarantee the documents answer the same question? Why or why not?

Answer:

No, it does not guarantee that the documents answer the same question.

Cosine similarity measures the similarity between two vectors, not whether their information is factually equivalent.

Two documents may discuss the same topic but contain different facts, dates, or conclusions.

I would evaluate the actual content and use a reranker or relevance evaluation to check whether the documents answer the query.

Key points: High similarity does not guarantee answer correctness.

## 11. If the chunk size is doubled, will the total number of stored vectors become exactly half? Explain your reasoning.

Answer:

No, not necessarily.

If the document length is fixed and there is no overlap, doubling the chunk size will approximately halve the number of chunks for long documents.

However, the exact number depends on document boundaries, splitting rules, overlap, and the length of each document.

Therefore, the number of vectors will not always be exactly half.

Key points: Approximation, document length, and overlap.

## 12. If chunk overlap increases from 0% to 50%, what changes would you expect in retrieval quality, storage requirements, and indexing costs?

Answer:

Increasing overlap can improve retrieval quality because important information near chunk boundaries is less likely to be lost.

However, it also creates more chunks with repeated content. This increases storage requirements, embedding costs, and indexing time.

Too much overlap can also return redundant results.

I would test different overlap values and select the one that provides the best balance between retrieval quality and cost.

Key points: Better boundary context, more storage, higher cost, and more redundancy.

## 13. If Top-K increases from 5 to 20, will retrieval accuracy always improve? Why or why not?

Answer:

No, increasing Top-K does not always improve retrieval accuracy.

A higher Top-K may retrieve more relevant information, but it can also introduce irrelevant chunks and redundant context.

This can confuse the LLM, increase token usage, and increase latency.

I would evaluate different Top-K values and use reranking to select the most relevant chunks.

Key points: Recall may improve, but precision and answer quality may decrease.
