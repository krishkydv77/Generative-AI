

## Part 1: RAG concepts and architecture

### Q1. What is RAG? Why would you choose it over fine-tuning? Can both approaches be combined?

Answer:

RAG stands for Retrieval-Augmented Generation. It retrieves relevant information from external documents and provides that information to an LLM to generate an accurate answer.

I choose RAG when I need answers based on private, updated, or domain-specific data. Fine-tuning is useful when I want to improve a model's behavior, style, or performance on a specific task.

Yes, both approaches can be combined. Fine-tuning can improve how the model responds, while RAG provides the latest and relevant information.

### Q2. Does a RAG system need both an embedding model and an LLM? What happens if you use only one?

Answer:

A typical RAG system uses both an embedding model and an LLM.

The embedding model converts text into numerical vectors so that we can find semantically similar documents. The LLM uses the retrieved information to generate a natural-language answer.

If I use only an embedding model, I can retrieve relevant documents, but I cannot generate a complete answer using an LLM.

If I use only an LLM, it can answer from its existing knowledge, but it cannot perform the usual embedding-based semantic retrieval.

### Q3. What is the difference between parametric knowledge and non-parametric knowledge?

Answer:

Parametric knowledge is the information learned by an LLM during training and stored in its model parameters.

Non-parametric knowledge is external information stored outside the model, such as in documents, databases, or vector databases.

For example, an LLM may know general Python concepts through its training. If I add my company's private HR policies to a vector database, the RAG system can retrieve those policies when answering questions.

The main difference is that external knowledge can be updated without retraining the LLM.

### Q4. How does query understanding affect retrieval quality?

Answer:

Query understanding helps the system understand the user's actual intention before retrieving documents.

If the query is unclear, too broad, or uses different words from the documents, the system may retrieve irrelevant information.

I can improve query understanding through query rewriting, query expansion, keyword extraction, and metadata filtering.

For example, if a user asks, "How can I get my money back?", the system can rewrite the query as "What is the refund policy?" to improve retrieval.

### Q5. What are the major failure points in a RAG pipeline? How would you identify the responsible component?

Answer:

The major failure points are document ingestion, text extraction, chunking, embedding generation, retrieval, reranking, context preparation, and LLM generation.

I would check each component separately using logs and evaluation metrics.

- If text is missing, I would check document extraction.
- If relevant information is split incorrectly, I would check chunking.
- If relevant documents are not retrieved, I would check embeddings and retrieval.
- If the correct documents are retrieved but the answer is wrong, I would check the prompt, context and LLM response.

I would use metrics such as Recall\@K, MRR, answer correctness, and faithfulness to identify the problem.

### Q6. Why can a RAG system produce incorrect answers even when the correct information exists in its knowledge base?

Answer:

A RAG system can produce incorrect answers because having the correct information in the database does not guarantee that it will be retrieved correctly.

The relevant information might be in a poorly created chunk, the embedding model might not match the query well, or the retriever might rank the wrong documents higher.

Even if the correct information is retrieved, the LLM may ignore it, misunderstand it, or generate unsupported details.

I would improve chunking, retrieval, reranking, prompt instructions, and answer evaluation.

### Q7. What is the difference between a RAG system and an Agentic RAG system?

Answer:

A traditional RAG system usually follows a fixed workflow: it retrieves relevant documents and sends them to an LLM to generate an answer.

Agentic RAG uses an AI agent that can make decisions during the process. It may rewrite queries, select different tools, retrieve information multiple times, or check whether the answer is sufficient.

For example, traditional RAG may perform one retrieval for a question. Agentic RAG may search multiple sources and retrieve additional information if the first result is insufficient.

Agentic RAG is more flexible but can be slower, more expensive, and more complex.

### Q8. When would you avoid Agentic RAG?

Answer:

I would avoid Agentic RAG when the task is simple, the workflow is predictable, or a normal RAG system already provides good results.

Agentic RAG introduces additional LLM calls, latency, cost, and debugging complexity.

For example, for a company FAQ chatbot that answers questions from a fixed set of documents, traditional RAG may be sufficient.

I would use Agentic RAG only when dynamic decisions, multiple tools, or iterative retrieval provide clear benefits.

### Q9. How would you decide whether a problem needs RAG, long-context prompting, fine-tuning, or a combination?

Answer:

I would choose the approach based on the data, task, update frequency, cost, and performance requirements.

- RAG: When information is private, frequently updated, or needs source-based answers.
- Long-context prompting: When I need to process a relatively small amount of information that fits within the model's context window.
- Fine-tuning: When I need to improve a model's behavior, response style, or performance on a specific task.
- Combination: When I need both specialized model behavior and access to updated external knowledge.

I would test the options on a representative dataset and compare answer quality, latency, and cost before choosing the final architecture.

### Q10. What is the difference between document-level retrieval and chunk-level retrieval?

Answer:

Document-level retrieval selects entire documents based on their relevance to the query.

Chunk-level retrieval selects smaller sections of documents that contain the most relevant information.

Document-level retrieval provides more context but may include unnecessary information. Chunk-level retrieval is more precise and often reduces the amount of text sent to the LLM.

In practice, I would retrieve relevant chunks and, when needed, include their parent document or neighboring chunks to provide additional context.

### Q11. How would you design a RAG architecture that supports frequently changing enterprise documents?

Answer:

I would design an incremental indexing pipeline.

First, I would ingest documents from sources such as SharePoint, internal databases, or cloud storage. Then I would extract the text, split it into chunks, generate embeddings, and store the vectors with metadata in a vector database.

I would track document versions, timestamps, and unique IDs. When a document changes, I would reprocess only the changed document or chunks instead of rebuilding the entire index.

I would also implement document deletion, access control, scheduled synchronization, monitoring, and version management.

This approach keeps the knowledge base updated while reducing processing cost and downtime.

## Part 2: RAG Cost Optimization Interview Questions

### Q1. How would you reduce the overall cost of a production RAG application without reducing answer quality?

Answer:

I would first measure the cost of each component to identify the main expense.

Then I would use several optimization techniques:

- Use smaller chunks with better retrieval quality.
- Use a smaller LLM for simple questions.
- Cache repeated queries and answers.
- Avoid generating embeddings for unchanged documents.
- Reduce unnecessary reranking and LLM calls.
- Use a local embedding model when it is cost-effective.

I would evaluate answer quality before and after every change to make sure the quality does not decrease.

### Q2. Which components contribute to RAG costs?

Answer:

The main cost components are embedding generation, vector database operations, reranking, LLM inference, document storage, and network operations.

LLM inference can be a major expense because we pay for input and output tokens when using a hosted API.

Embedding costs mainly occur during indexing and when processing new queries. Vector database costs depend on storage, search volume, and infrastructure.

I would monitor each component separately to identify where the budget is being spent.

### Q3. What is the difference between one-time indexing cost and recurring query-time cost?

Answer:

Indexing cost is the cost of preparing documents for retrieval. It includes text extraction, chunking, embedding generation, and storing vectors.

Query-time cost occurs whenever a user asks a question. It includes query embeddings, retrieval, reranking, and LLM generation.

Indexing costs are mostly incurred when documents are first processed or updated. Query-time costs continue as users interact with the application.

I would optimize indexing through incremental updates and query-time costs through caching and efficient model selection.

### Q4. How would you decide whether to use a small, inexpensive LLM or a more powerful, expensive LLM in a RAG system?

Answer:

I would compare both models using a representative set of questions.

I would evaluate answer accuracy, faithfulness, latency, token usage, and cost.

For simple questions, I would use a smaller and cheaper model. For complex questions that require reasoning or combining information from multiple documents, I would use a more powerful model.

I could also implement model routing, where a small model handles most queries and difficult queries are sent to a stronger model.

### Q5. Your RAG application handles 1 lakh queries per day. How would you estimate the monthly LLM cost?

Answer:

First, I would calculate the monthly query volume.

One lakh queries per day means approximately 30 lakh queries per month, assuming 30 days.

Then I would estimate the average input and output tokens per query and multiply them by the model's token prices.

For example, suppose each query uses 2,000 input tokens and 300 output tokens.

- Monthly input tokens: 6 billion.
- Monthly output tokens: 900 million.

The estimated monthly cost would be:

\\[ \begin{aligned} \text{Cost} ={}& \frac{6{,}000{,}000{,}000}{1{,}000{,}000} \times P\_{\text{input}}\\\ &+\frac{900{,}000{,}000}{1{,}000{,}000} \times P\_{\text{output}} \end{aligned} \\]

Here, \\(P\_{\text{input}}\\) and \\(P\_{\text{output}}\\) are the prices per million tokens.

I would then account for caching, retries, model routing, and other infrastructure costs. The actual cost depends on the selected model and its current pricing.

### Q6. How would you reduce embedding API costs when ingesting millions of document chunks?

Answer:

I would use batch processing wherever the embedding provider supports it.

I would remove duplicate documents, avoid embedding unchanged chunks, and use incremental indexing.

I would also evaluate a smaller or local embedding model if it provides sufficient retrieval quality.

For large datasets, I would process documents in batches, use checkpoints, and implement retries so that failed batches do not need to be processed again.

### Q7. When should you use a local embedding model instead of a hosted embedding API?

Answer:

I would consider a local embedding model when the application processes a large number of documents, has strict data privacy requirements, or needs predictable costs.

A hosted API is easier to manage because the provider handles the infrastructure.

A local model can reduce per-request API charges, but it requires computing resources, maintenance, and monitoring.

I would compare the total operating cost, retrieval quality, speed, and hardware requirements before deciding.

### Q8. How would you implement semantic caching to reduce repeated LLM calls? What are the risks?

Answer:

Semantic caching stores previous queries, their embeddings, and their answers.

When a new query arrives, the system generates its embedding and compares it with cached query embeddings. If a sufficiently similar query exists, it can return the cached answer instead of calling the LLM again.

I would use a similarity threshold and validate that the cached result is appropriate for the current user and context.

The main risks are returning outdated answers, treating similar but different questions as identical, and leaking private information between users.

To reduce these risks, I would use access-control checks, cache expiration, document-version validation, and carefully tested similarity thresholds.

### Q9. How would you cache retrieved chunks, generated answers, or both?

Answer:

I would use different caching strategies for different stages.

- Retrieved-chunk cache: Stores retrieval results for repeated or similar queries. It reduces repeated retrieval work.
- Answer cache: Stores previously generated answers. It can avoid both retrieval and LLM generation when a valid cached answer can be reused.
- Both: Can reduce costs across the retrieval and generation stages.

I would invalidate cached results when source documents change. For private enterprise data, I would also ensure that cached content is never shared across users who have different access permissions.

### Q10. How can incremental indexing reduce costs compared with rebuilding the entire vector index?

Answer:

Incremental indexing processes only new, modified, or deleted documents.

Without incremental indexing, the system may need to extract, chunk, and embed every document again, even if most documents have not changed.

With incremental indexing, I would track document IDs, versions, and content hashes. I would compare the latest documents with the existing index and update only the affected chunks.

This reduces embedding API costs, processing time, compute usage, and unnecessary database operations.

### Q11. How would you identify whether your RAG application is spending too much on retrieval, generation, or infrastructure?

Answer:

I would monitor costs, latency, token usage, and resource utilization for each component.

For retrieval, I would check vector database query costs, query latency, and reranking overhead.

For generation, I would monitor input tokens, output tokens, model prices, and the number of LLM calls.

For infrastructure, I would monitor CPU, memory, storage, network traffic, and idle resources.

I would also calculate cost per query and cost per successful answer. This helps me identify expensive components and optimize them without sacrificing quality.
