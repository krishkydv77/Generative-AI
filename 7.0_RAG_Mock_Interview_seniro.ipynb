# RAG Mock Interview — Questions & Answers


## Part 1 — RAG Fundamentals

### Q1. What is RAG?

Answer:

RAG stands for Retrieval-Augmented Generation. It is a technique that retrieves relevant information from external documents and provides that information to an LLM to generate an answer.

It helps the model answer questions using specific data and reduces hallucinations.

### Q2. Why do we use RAG?

Answer:

We use RAG when an LLM needs information that may not be available in its training data.

For example, a company chatbot can answer questions using internal company documents without training a new model.

### Q3. How does RAG work?

Answer:

RAG works in two main phases: indexing and retrieval.

1. First, we load documents.
2. We split documents into chunks.
3. We convert chunks into embeddings.
4. We store embeddings in a vector database.
5. When a user asks a question, we convert the question into an embedding.
6. We retrieve relevant chunks.
7. We send the question and retrieved context to the LLM.
8. The LLM generates the final answer.

### Q4. What is the difference between RAG and fine-tuning?

Answer:

RAG provides external information to the model at query time. Fine-tuning trains a model further on a specific dataset.

RAG is useful for frequently changing information. Fine-tuning is useful for adapting a model's behavior, style, or task-specific responses.

### Q5. What is an embedding?

Answer:

An embedding is a list of numbers that represents the meaning of text.

Texts with similar meanings usually have similar embeddings. We use embeddings to find relevant documents based on semantic similarity.

### Q6. What is a vector database?

Answer:

A vector database stores embeddings and helps us search for similar vectors efficiently.

Examples include FAISS, Chroma, Pinecone, and Qdrant.

### Q7. What is chunking in RAG?

Answer:

Chunking means splitting large documents into smaller pieces of text.

We use chunking because sending an entire document to the model can be expensive and may exceed the context window. Good chunking improves retrieval quality.

### Q8. What is chunk overlap?

Answer:

Chunk overlap means keeping some text from one chunk in the next chunk.

It helps preserve context between chunks and reduces the chance of losing important information at chunk boundaries.

### Q9. What is semantic search?

Answer:

Semantic search finds information based on meaning rather than only exact keywords.

For example, a search for "money-back procedure" may retrieve a document titled "Refund Policy" if their meanings are similar.

### Q10. What is hallucination in an LLM?

Answer:

Hallucination happens when an LLM generates incorrect or unsupported information.

RAG can reduce hallucinations by providing relevant context, but it cannot eliminate them completely. We should also use reliable retrieval, clear prompts, and proper evaluation.

## Part 2 — Technical Concepts

### Q11. What is the difference between keyword search and vector search?

Answer:

Keyword search finds documents using exact words or matching terms. Vector search finds documents using semantic similarity between embeddings.

For better results, we can combine both methods using hybrid search.

### Q12. What is cosine similarity?

Answer:

Cosine similarity measures how similar two vectors are based on the angle between them.

A higher score generally means the vectors point in more similar directions. It is commonly used to compare query and document embeddings.

### Q13. What is a retriever in RAG?

Answer:

A retriever finds the most relevant document chunks for a user's query.

It searches the document collection and returns the chunks that will be passed to the LLM as context.

### Q14. What is Top-K retrieval?

Answer:

Top-K retrieval returns the K most relevant chunks for a query.

For example, if K is 5, the retriever returns the five highest-ranked chunks. We need to choose K carefully because too few chunks may miss information, while too many may add noise.

### Q15. What is the role of the LLM in RAG?

Answer:

The LLM uses the retrieved context and the user's question to generate a natural-language answer.

The retriever finds information, while the LLM uses that information to formulate the response.

### Q16. What is a prompt template in RAG?

Answer:

A prompt template is a predefined structure for sending instructions, context, and the user's question to an LLM.

For example, we can instruct the model to answer only from the provided context and say "I don't know" when the information is missing.

### Q17. What is metadata filtering?

Answer:

Metadata filtering restricts retrieval using additional information about documents.

For example, we can retrieve only documents from the HR department or documents published in 2026. This can improve relevance and access control.

### Q18. What is hybrid search?

Answer:

Hybrid search combines keyword search and vector search.

Keyword search is useful for exact terms, product IDs, and names. Vector search is useful for semantic meaning. Combining both can improve retrieval quality.

### Q19. What is a reranker?

Answer:

A reranker takes the initial retrieved results and scores them again to improve their order.

For example, the retriever may return 20 chunks, and the reranker selects the most relevant 5 chunks for the LLM.

### Q20. How do you choose the chunk size?

Answer:

I choose the chunk size based on the document type, the structure of the content, and the retrieval results.

I start with a reasonable chunk size and overlap, test different configurations, and evaluate retrieval quality. There is no single best chunk size for every dataset.

## Part 3 — Scenario-Based Questions


### Q21. Your RAG system retrieves irrelevant documents. How would you debug it?

Answer:

First, I would check the quality of the documents and chunking.

Then I would verify the embedding model, query embedding, and similarity metric. I would also check the Top-K value and retrieval scores.

Finally, I would test hybrid search or a reranker and evaluate the results using sample queries.

### Q22. The correct document exists, but retrieval fails because the query uses different words. What would you do?

Answer:

This can happen when keyword search cannot match similar meanings.

I would use semantic search or hybrid search. I would also check the embedding model and test whether query rewriting improves retrieval.

### Q23. How do you reduce hallucinations in RAG?

Answer:

I would retrieve high-quality context, use a clear prompt, and instruct the LLM to answer only from the provided information.

If the context does not contain the answer, the model should say that the information is unavailable. I would also evaluate answers against trusted references.

### Q24. What happens if the retrieved context exceeds the LLM's context window?

Answer:

I would reduce unnecessary chunks and select only the most relevant information.

I could also use reranking, context compression, or a smaller Top-K value. The goal is to provide useful context without exceeding the model's token limit.

### Q25. What is the difference between precision and recall in retrieval?

Answer:

Precision measures how many retrieved documents are relevant.

Recall measures how many of all relevant documents were successfully retrieved.

High precision means less irrelevant information. High recall means fewer relevant documents are missed. We need to balance both based on the application.

### Q26. How do you evaluate a RAG system?

Answer:

I evaluate both retrieval and generation.

For retrieval, I use metrics such as Precision\@K, Recall\@K, and MRR.

For generation, I check answer correctness, faithfulness to the retrieved context, relevance, and whether the answer contains unsupported claims.

### Q27. What is the difference between RAGAS and traditional retrieval metrics?

Answer:

Traditional retrieval metrics measure how well the retriever finds relevant documents.

RAGAS provides a framework for evaluating RAG pipelines with metrics such as faithfulness, answer relevancy, and context precision or recall, depending on the evaluation setup.

### Q28. How would you handle PDF documents in a RAG application?

Answer:

I would extract text from PDFs using tools such as PyMuPDF or pdfplumber.

Then I would clean the extracted text, split it into chunks, generate embeddings, and store them in a vector database.

For scanned PDFs, I would use OCR, such as Tesseract, to extract text from images.

### Q29. How would you make a RAG application production-ready?

Answer:

I would build an API using FastAPI, validate inputs, and handle errors properly.

I would add authentication, access control, logging, monitoring, and caching where appropriate. I would also evaluate retrieval quality, manage API costs, and use Docker for deployment.

### Q30. What are the limitations of RAG?

Answer:

RAG depends on the quality of the source documents and retrieval results.

If the retriever finds incorrect or incomplete context, the LLM may generate a poor answer. RAG can also increase latency and require additional storage and evaluation.

We need to improve document quality, retrieval, and generation together.
