# What is chunking in RAG?
**ans.** chunking is the the process of spliting a large documents into smaller manageable pieces of text called chunks before storing them in vectors database.

## chunking important in RAG:-
- batter retrieval
- improve relevance
-  context management
- reduce noise
-  embedding efficiency
-  cost optimization
## parameter;
- **chunk size**:- maximum or target size of a chunk ,measured in charactares or tekons.
- **chunk overlap**:- text repeated b/w consecutive chunks to preserve context.
- **separators** :- boundaries used to split text ,such as paragraph,newlines,pr sapce.
- **Add start index**:- store each chunk starting position in the original source text.

# Types of Chunking:-
## **1. Fixed-Size Chunking**
It splits text into chunks of a predefined size, usually measured in characters or tokens.
- **Advantages:** Simple, fast, and easy to implement.
- **Limitation:** May split sentences or related information.
- **Use Case**: Simple, consistently formatted text.
Example: Split a document into chunks of 500 characters each.
## **2. Recursive Character Chunking**
It progressively uses smaller separators, such as paragraphs, newlines, and spaces, to split text while preserving natural boundaries wherever possible.
- **Advantages:** Preserves paragraphs and sentences better than basic fixed-size splitting.
- **Limitation:** Chunk boundaries may still split related ideas.
- **Use Case:** General-purpose RAG applications, articles, and documentation.
## **3. Document Structure Chunking**
It splits documents according to their existing structure, such as headings, sections, paragraphs, pages, or Markdown headers.
- **Advantages:** Preserves the document's logical organization.
- **Limitation:** Requires document structure to be identified correctly.
- **Use Case:** PDFs, manuals, policies, and Markdown documents.
## **4. Semantic Chunking**
It groups text based on changes in meaning or semantic similarity, often using embeddings.
- **Advantages:** Keeps related ideas together and can improve retrieval for some datasets.
- **Limitation:** Requires additional computation and tuning.
- **Use Case:** Documents containing complex concepts or changing topics.
## **5. Parent-Child Chunking**
It retrieves smaller child chunks but can provide their larger parent section as context to the LLM.
- **Advantages:** Small child chunks improve retrieval precision, while larger parent sections preserve context.
- **Limitation:** Requires maintaining relationships between child and parent chunks.
- **Use Case:** Complex documents, technical manuals, and detailed knowledge bases.

## 3. Example: LangChain se Chunking

Ye RAG projects mein commonly used recursive chunking ka example hai.

```
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """Chunking is important in RAG.It splits large documents into smaller pieces.These chunks are converted into embeddingsand stored in a vector database."""

splitter = RecursiveCharacterTextSplitter(    chunk_size=100,    chunk_overlap=20,    length_function=len)

chunks = splitter.split_text(text)

for i, chunk in enumerate(chunks, start=1):
    print(f"Chunk {i}: {chunk}")
```

# Chunking in RAG — Interview Questions & Answers

### 1. What is the difference between chunk size and chunk overlap?

Answer: Chunk size is the length of each text chunk, while chunk overlap is the text repeated between two consecutive chunks.

Example: If the chunk size is 500 characters and the overlap is 50 characters, the last 50 characters of one chunk are repeated in the next chunk.

### 2. What happens if chunk size is too small or too large?

Answer: If the chunk size is too small, important context may be lost. If it is too large, irrelevant information may be included, reducing retrieval quality and increasing token usage.

### 3. How do you choose the best chunk size?

Answer: I choose the chunk size based on document type, query complexity, and the LLM's context window. I test different chunk sizes and compare retrieval quality to find the best one.

### 4. Can chunking affect RAG accuracy?

Answer: Yes, chunking directly affects retrieval quality. Good chunking retrieves relevant information, while poor chunking may break important context and lead to incorrect answers.

### 5. How do you evaluate a chunking strategy?

Answer: I test it using sample questions and check whether the correct chunks are retrieved. I evaluate metrics such as Precision\@K, Recall\@K, and answer correctness.

### 6. How do you chunk PDFs containing tables?

Answer: I extract tables using tools like PyMuPDF or pdfplumber and preserve their rows, columns, and headers. I keep related table information together so the meaning is not lost.

### 7. How do you prevent duplicate information in chunks?

Answer: I use controlled overlap, remove duplicate text during preprocessing, and check for duplicate chunks using hashes or similarity checks.

### 8. If a document changes every week, how would you update its chunking?

Answer: I detect document changes, reprocess the updated content, regenerate embeddings for changed chunks, and update the vector database. I also remove outdated chunks to prevent stale information.

### 9. How do you optimize chunking for retrieval quality?

Answer: I experiment with chunk size, overlap, and splitting methods. I evaluate retrieval results on sample queries and select the strategy that provides the most relevant context.

### 10. If the retriever returns the correct document but the final answer is wrong, how would you determine whether chunking is the cause?

Answer: I inspect the retrieved chunks to see whether they contain enough information to answer the question. If important context is missing or split across chunks, chunking may be the problem. Otherwise, I check the prompt, retrieved context, and LLM response.

### 11. Why is token-based chunking sometimes preferred?

Answer: Token-based chunking is preferred because LLMs process input in tokens, not characters. It helps control input length and manage the model's context window more accurately.

### 12. Is 500 characters always the best chunk size?

Answer: No, 500 characters is not always the best size. The ideal chunk size depends on the document, the type of questions, and retrieval performance.

### 13. Does overlap always improve retrieval?

Answer: No, overlap does not always improve retrieval. It helps preserve context between chunks, but excessive overlap can create duplicate information and increase storage and retrieval costs.

### 14. Difference between chunk size and chunk overlap?

Answer: Chunk size defines how much text a chunk contains, while chunk overlap defines how much text is shared between consecutive chunks.

### 15. What is chunking, and why is it important in RAG?

Answer: Chunking is the process of splitting large documents into smaller pieces of text before storing their embeddings in a vector database. It helps retrieve relevant information, manage context, reduce irrelevant text, and improve the quality of RAG responses.

````python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("/content/novel.txt")
docs = loader.load()


print(f"Total characters in novel: {len(docs[0].page_content)}")
````

````python
from langchain_text_splitters import CharacterTextSplitter

# 1. Fixed-Size Chunking
fixed_splitter = CharacterTextSplitter(
    separator="\n", #Text ko newline ke basis par split karega.
    chunk_size=500, #Har chunk max 500 characters ka hoga.
    chunk_overlap=50, #Previous chunk ke 50 characters next chunk me repeat honge.
    length_function=len #Character count calculate karne ke liye.
)

fixed_chunks = fixed_splitter.split_documents(docs) #Documents ko chunks me tod diya.
print(f"Fixed-Size Chunks Total: {len(fixed_chunks)}") #Total chunks count kar raha hai.
print("Sample Chunk:", fixed_chunks[0].page_content[:500]) #Pehla chunk dikha raha hai.
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 2. Recursive Character Chunking
recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    separators=["\n\n", "\n", " ", ""] #Pehle paragraph level par split karega./Line level par split karega./Space par split karega./Character level par split karega.
)

recursive_chunks = recursive_splitter.split_documents(docs)
print(f"Recursive Chunks Total: {len(recursive_chunks)}")
print("Sample Chunk:", recursive_chunks[0].page_content[:500])
````

````python
import re #Regex use karne ke liye.
from langchain_core.documents import Document #Custom Document objects banane ke liye.

# 3. Document-Structure Chunking (Splitting by Chapters)
text_content = docs[0].page_content
# Regex to match 'Chapter [Number]' case-insensitively
chapters = re.split(r'\n(?=Chapter\s+\d+|CHAPTER\s+[IVXLCDM]+)', text_content, flags=re.IGNORECASE)

structural_docs = [
    Document(page_content=ch.strip(), metadata={"source": "novel.txt", "chapter": i})
    for i, ch in enumerate(chapters) if ch.strip() #Har chapter ko number assign. and Empty chapter skip.
]

print(f"Document-Structure Chunks (Chapters) Total: {len(structural_docs)}")
if structural_docs:
    print("First Chapter Preview:", structural_docs[0].page_content[:300])
````

````python
from langchain_experimental.text_splitter import SemanticChunker #Meaning based chunking.
from langchain_community.embeddings import HuggingFaceEmbeddings #Embeddings generate karne ke liye.

# Local embedding model load karein (Free & Fast)
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2") #Sentence ko vectors me convert karega.

# 4. Semantic Chunking
semantic_splitter = SemanticChunker(   #Semantic similarity ke basis par split karega. means Jaha topic change hoga waha chunk break hoga.
    embeddings,
    breakpoint_threshold_type="percentile" # ya standard_deviation
)

semantic_chunks = semantic_splitter.split_documents(docs)
print(f"Semantic Chunks Total: {len(semantic_chunks)}")
if semantic_chunks:
    print("Sample Semantic Chunk:", semantic_chunks[0].page_content[:200])
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_core.documents import Document

# 1. Embeddings aur Vector Store setup
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vectorstore = FAISS.from_texts([""], embedding=embeddings)

# 2. Parent aur Child Splitters define karein
parent_splitter = RecursiveCharacterTextSplitter(chunk_size=1500, chunk_overlap=150)
child_splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=50)

# 3. Manual Parent-Child Chunking Implementation (Bina kisi missing import ke)
parent_docs = parent_splitter.split_documents(docs)

child_docs = []
for i, parent_doc in enumerate(parent_docs): #Har parent chunk par loop.
    # Har parent chunk ke andar chhote child chunks banayein
    small_chunks = child_splitter.split_documents([parent_doc]) #Parent ko aur chhote chunks me tod diya.

    for small_chunk in small_chunks: #Har child chunk par loop.
        # Child metadata me parent ki ID ya link jod dein
        small_chunk.metadata["parent_id"] = i #Child ko parent se link kar diya.
        child_docs.append(small_chunk)

# Vector store me sirf chote (child) chunks dalein taaki search fast aur accurate ho
vectorstore.add_documents(child_docs) #Sab child chunks FAISS me store.

print(f"Parent Chunks Total: {len(parent_docs)}")
print(f"Child Chunks Total: {len(child_docs)}")
print("Parent-Child Chunking Setup Successfully Completed!")
````
